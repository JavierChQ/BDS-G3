# Funciones para procesamiento de datos

Limpieza de valores nulos

```python
import numpy as np
df_books['Price'] = np.nan
df_books.isnull().value_counts()
```

```python
df_books.loc[df_books['Year'] < 2016,'Price'] = 1000
df_books.isnull().sum()
```

```python
df_books.dropna()
```

```python
df_books.fillna(0)
```

# Limpieza de valores duplicados

```python
df_books.iloc[0]
```

```python
df_books = pd.concat([df_books,df_books.tail(3)],ignore_index= True)
df_books
```

```python
df_books.drop_duplicates(keep='last',inplace=True)
df_books
```
